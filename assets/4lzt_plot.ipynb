```python
%matplotlib inline
import matplotlib
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Pepsi (AA)
scaling = 9.14376e-08  # "Intensity Scaling" from .log file
q, Itotal, Iatomic, Iexcluded, Ihydration, AatAev, AatAhs, AevAhs = np.loadtxt(
    "4lzt_pepsi.out", unpack=True
)
plt.plot(q, Itotal / scaling, label="Pepsi (AA)")

# Pripps (CG)
q, I = np.loadtxt("4lzt.cg_intensity.dat", unpack=True)
plt.plot(q, I, label="Pripps (CG, no solvation)")

# Experiment
q, I, b = np.loadtxt("SASDMF2.dat", unpack=True)
plt.plot(q, I * 0.7e5, label="Experiment (SASDMF2)")

# Plot details
plt.title("Calculated and measured formfactor for Lysozyme")
plt.yscale("log")
plt.legend(loc=0, frameon=False)
plt.xlim(0, 0.75)
plt.xlabel("q (1/Å)")
plt.ylabel("Intensity")
plt.show()
```
