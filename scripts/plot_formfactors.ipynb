```python
%matplotlib inline
import matplotlib
import numpy as np
import matplotlib.pyplot as plt
import yaml
```

```python
with open("formfactors.yaml") as file:
    data = yaml.load(file, Loader=yaml.FullLoader)
    data = data["H2O"]
    q = [pair[0] for pair in data]
    I = [pair[1] for pair in data]
    plt.plot(q, I, lw=3, alpha=0.5)
    plt.yscale("log")
```

```python
import mdtraj as md
traj = md.load("tests/4lzt/4lzt.pdb")
sasa = md.shrake_rupley(traj, mode="residue")[0] * 100.0 # -> Å^2
print(f"Total SASA = {sasa.sum():.2f} Å^2 (N={len(sasa)})")
print(sasa)
np.savetxt("4lzt_sasa.dat", sasa, header="Residue SASA for 4lzt.pdb from mdtraj.shrake_rupley()")
```
